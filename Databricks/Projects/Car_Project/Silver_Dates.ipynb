```python
from datetime import date, timedelta
import pandas as pd
from pyspark.sql.functions import expr, monotonically_increasing_id
from pyspark.sql.types import DateType

# Step 1: Generate date range using pandas
start_date = date(2010, 1, 1)
end_date = date(2026, 1, 1)
date_range = pd.date_range(start=start_date, end=end_date)

# Step 2: Create Spark DataFrame
df_dates = spark.createDataFrame(pd.DataFrame({'Date': date_range}))

# Step 3: Add calendar columns
df_dates = df_dates \
    .withColumn("Date_ID", expr("date_format(Date, 'yyyyMMdd')")) \
    .withColumn("Day", expr("day(Date)")) \
    .withColumn("Month", expr("month(Date)")) \
    .withColumn("Year", expr("year(Date)")) \
    .withColumn("DayOfWeek", expr("dayofweek(Date)")) \
    .withColumn("WeekOfYear", expr("weekofyear(Date)")) \
    .withColumn("Quarter", expr("quarter(Date)"))\
    .withColumnRenamed("Date_ID", "DateKey")

# Step 4: Add surrogate key (DimKey)

# Step 5: Reorder columns with DimKey first
df_dates = df_dates.select("DateKey","Date", "Day", "Month", "Year", "DayOfWeek", "WeekOfYear", "Quarter")

# Step 6: Save to Delta table
df_dates.write.format("delta").mode("overwrite").saveAsTable("Car_Project.Silver.Dates")
```
