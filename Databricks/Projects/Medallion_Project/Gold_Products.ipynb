
# DLT Piplines

```python
import dlt
from pyspark.sql.functions import *
```

### Streaming table

```python
@dlt.table()
   
def DimProducts_stage():
    df = spark.readStream.table("databricks_cata.silver.products_silver")

    return df
```

### Streaming view
Materialised view not only saves the query it also stores the result of the query in a table

```python
@dlt.view()

def Dimproducts_view():
    df = spark.readStream.table("Live.DimProducts_stage")
    return df
```

**Dim Products**

```python
dlt.create_streaming_table("DimProducts")
```

```python
dlt.apply.changes(
    target = "databricks_cata.silver.Dimproducts",
    source = "dlt.Dimproducts_view",
    keys = ["product_id"],
    sequence_by = "product_id",
    stored_as_scd_type = 2
)
```
