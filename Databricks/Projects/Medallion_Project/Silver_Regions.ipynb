```python

# Reading from a table instead a Data Lake

df = spark.read.table("databricks_cata.bronze.regions")
```

```python
df = df.drop("_rescued_data")
df.display()
```

```python

# SAVING THE TABLE FROM THE BRONZE LAYER IN DELTA FORMAT IN THE SILVER LAYER

df.write.format("delta")\
    .mode("overwrite")\
    .save("abfss://silver@databricksuk2025.dfs.core.windows.net/regions")
```

```python
%sql

-- CREATING THE SILVER TABLE
create table if not exists databricks_cata.silver.regions_silver 
using DELTA
Location 'abfss://silver@databricksuk2025.dfs.core.windows.net/regions'
```
